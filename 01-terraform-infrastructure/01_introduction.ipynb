## What is Infrastructure as Code?

**Infrastructure as Code (IaC)** is the practice of managing and provisioning infrastructure through machine-readable configuration files rather than manual processes.

### Traditional Approach vs IaC

| Traditional | Infrastructure as Code |
|------------|------------------------|
| Click through Dashboard | Write configuration files |
| Manual, error-prone | Automated, consistent |
| Hard to replicate | Easy to replicate |
| No version history | Full version control |
| Difficult to audit | Clear audit trail |

## Why Use Terraform with Stripe?

### Key Benefits

1. **Version Control**: Track all changes to your Stripe configuration in Git
2. **Reproducibility**: Spin up identical environments (dev, staging, prod)
3. **Collaboration**: Review Stripe configuration changes through pull requests
4. **Automation**: Integrate with CI/CD pipelines
5. **Documentation**: Your configuration IS your documentation

### Use Cases

- **Product Catalog Management**: Define products and prices as code
- **Multi-Environment Setup**: Same products in test and live modes
- **Compliance**: Audit trail for all configuration changes
- **Disaster Recovery**: Quickly recreate your Stripe setup

## The Stripe Terraform Provider

The [Stripe Terraform Provider](https://registry.terraform.io/providers/stripe/stripe/latest/docs) is the official provider maintained by Stripe.

### Some of the Supported Resources

| Resource | Description |
|----------|-------------|
| `stripe_product` | Products in your catalog |
| `stripe_price` | Prices for products (one-time or recurring) |
| `stripe_customer` | Customer records |
| `stripe_coupon` | Discount coupons |
| `stripe_webhook_endpoint` | Webhook configurations |
| `stripe_portal_configuration` | Customer portal settings |

### Provider Configuration

```hcl
terraform {
  required_providers {
    stripe = {
      source  = "stripe/stripe"
      version = "0.1.3"
    }
  }
}

provider "stripe" {
  # API key is read from STRIPE_API_KEY environment variable
}
```

## Architecture Overview

```
+------------------+      +------------------+      +------------------+
|                  |      |                  |      |                  |
|  Terraform CLI   | ---> |  Stripe Provider | ---> |   Stripe API     |
|                  |      |                  |      |                  |
+------------------+      +------------------+      +------------------+
        |                                                    |
        v                                                    v
+------------------+                              +------------------+
|  .tf config      |                              |  Stripe Dashboard|
|  files           |                              |  (view results)  |
+------------------+                              +------------------+
```

### Workflow

1. **Write**: Define resources in `.tf` files
2. **Plan**: Preview what Terraform will create/modify/delete
3. **Apply**: Execute the changes against Stripe API
4. **Verify**: Check results in Stripe Dashboard

## Example: A Simple Product

Here's what a product definition looks like in Terraform:

```hcl
resource "stripe_product" "pro_plan" {
  name        = "Pro Plan"
  description = "Professional tier with advanced features"
}

resource "stripe_price" "pro_monthly" {
  product     = stripe_product.pro_plan.id
  currency    = "usd"
  unit_amount = 2900  # $29.00
  recurring {
    interval = "month"
  }
}
```

This creates:
- A product named "Pro Plan"
- A monthly recurring price of $29
