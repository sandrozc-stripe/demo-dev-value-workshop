## Step 1: Explore the Configuration

````python
# View the provider configuration
!cat main.tf
````

### Understanding main.tf

The `main.tf` file configures the Stripe provider:

```hcl
terraform {
  required_providers {
    stripe = {
      source  = "stripe/stripe"  # Official Stripe provider
      version = "0.1.3"          # Provider version
    }
  }
}

provider "stripe" {
  # API key is automatically read from STRIPE_API_KEY env var
}
```

````python
# View the product configuration
!cat products.tf
````

### Understanding products.tf

This file defines:

1. **A Product** (`stripe_product.pro_plan`)
   - Name: "Pro Plan"
   - Description: "Professional tier with advanced features"

2. **Two Prices**:
   - Monthly: $29/month (`stripe_price.pro_monthly`)
   - Yearly: $300/year (`stripe_price.pro_yearly`)

Note: `unit_amount` is in **cents** (2900 = $29.00)

## Step 2: Initialize Terraform

## Step 3: Plan the Changes

The `terraform plan` command shows what Terraform will create **without actually creating it**.

This is a **dry run** - safe to execute anytime!

<!-- PRESENTER: Emphasize the importance of planning before applying -->

````python
# Preview what Terraform will create
!terraform plan
````

### Understanding the Plan Output

The plan shows:
- **`+`** Resources to be created
- **`-`** Resources to be destroyed
- **`~`** Resources to be modified

You should see:
- 1 product to create
- 2 prices to create
- 1 customer to create

**Total: 4 resources to add**

## Step 4: Apply the Configuration

Now let's create the resources in Stripe!

The `-auto-approve` flag skips the confirmation prompt (use with caution in production).

<!-- PRESENTER: Show the Stripe Dashboard before and after applying -->

````python
# Apply the configuration (creates resources in Stripe)
!terraform apply -auto-approve
````

### Checkpoint

You should see:
- `Apply complete! Resources: 4 added, 0 changed, 0 destroyed.`
- Output values showing the created price IDs

**Dashboard**: Navigate to [Products](https://dashboard.stripe.com/test/products) to see your "Pro Plan" product!
Tips: you can use the Stripe Dashboar Search bar and paste the id

## Step 5: View Outputs

````python
# View the output values
!terraform output
````

## Step 6: View State

Terraform tracks created resources in a **state file**. This is how it knows what exists.

````python
# List resources in state
!terraform state list
````

````python
# Show details of a specific resource
!terraform state show stripe_product.pro_plan
````
